```python
import os
import shutil
from tqdm import tqdm
```

```python
DATA_DIR = "../Datasets/"
OUT_DIR = "../Datasets/Processed_Data"

views = ["Front", "Lateral"]
```

```python
# Create Processed_Data folder structure
for view in views:
    os.makedirs(os.path.join(OUT_DIR, view), exist_ok=True)
```

```python


def process_folder(view):
    src_path = os.path.join(DATA_DIR, view)
    out_view_path = os.path.join(OUT_DIR, view)

    files = [f for f in os.listdir(src_path) if f.endswith(".mp4")]

    for f in tqdm(files, desc=f"Processing {view}"):
        word_code = f[:4]  # 'W001', 'W045', etc.
        word_dir = os.path.join(out_view_path, word_code)
        os.makedirs(word_dir, exist_ok=True)

        shutil.copy(
            os.path.join(src_path, f),
            os.path.join(word_dir, f)
        )

```

```python
# Process both views
for v in views:
    process_folder(v)

print("Folder processing complete!")
```
