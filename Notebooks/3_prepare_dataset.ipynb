```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
PROC_DIR = "../Datasets/normalized_landmarks"
```

```python
samples = []
```

```python
def get_label_string_from_dir(dir_name):
    # dir_name will be 'W001', 'W002', etc. We return it as is.
    return dir_name
```

```python
# Use os.walk to get all files recursively
for dirpath, dirnames, filenames in os.walk(PROC_DIR):
    for f in filenames:
        if f.endswith(".npy"):
            # Get the name of the immediate parent directory (e.g., 'W001')
            parent_dir_name = os.path.basename(dirpath)
            
            # Use the new function to get the label string
            label = get_label_string_from_dir(parent_dir_name)
            
            full_path = os.path.join(dirpath, f)
            samples.append([full_path, label])
```

```python
df = pd.DataFrame(samples, columns=["file", "label"])
```

```python
print(f"DataFrame shape after loading: {df.shape}")
print(f"Unique labels found: {df.label.unique()}") # Check the labels
```

```python
# Split 70/15/15
# The rest of your code remains the same
train_df, test_df = train_test_split(df, test_size=0.15, stratify=df.label)
train_df, val_df = train_test_split(train_df, test_size=0.15, stratify=train_df.label)
```

```python
train_df.to_csv("../datasets/train.csv", index=False)
val_df.to_csv("../datasets/val.csv", index=False)
test_df.to_csv("../datasets/test.csv", index=False)
```
